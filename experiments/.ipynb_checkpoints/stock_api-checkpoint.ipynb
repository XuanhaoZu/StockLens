```python
import requests
import pandas as pd

# Replace 'demo' with your own Alpha Vantage API key
API_KEY = '5280A66MWDO73OD0'
BASE_URL = 'https://www.alphavantage.co/query'
SYMBOL = 'AAPL'

def fetch_stock_data(symbol):
    # Set up the parameters for the API request
    parameters = {
        'function': 'TIME_SERIES_DAILY',
        'symbol': symbol,
        'apikey': API_KEY
    }
    
    # Make the HTTP GET request to the API
    response = requests.get(BASE_URL, params=parameters)
    data = response.json()
    
    # Extract the time series data
    if 'Time Series (Daily)' not in data:
        raise ValueError('Invalid API response. Check your symbol and API key.')
        
    ts_data = data['Time Series (Daily)']
    
    # Convert the JSON data into a pandas DataFrame for easier analysis
    df = pd.DataFrame.from_dict(ts_data, orient='index')
    df.columns = ['Open', 'High', 'Low', 'Close', 'Volume']
    df.index = pd.to_datetime(df.index)
    df = df.astype(float)
    
    return df

# Fetch and print the data
try:
    stock_df = fetch_stock_data(SYMBOL)
    print(stock_df.head())
except Exception as e:
    print(f"Error encountered: {e}")
```
