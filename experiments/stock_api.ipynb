```python
import requests
import pandas as pd

# Replace 'demo' with your own Alpha Vantage API key
API_KEY = '5280A66MWDO73OD0'
BASE_URL = 'https://www.alphavantage.co/query'
SYMBOL = 'AAPL'

def fetch_stock_data(symbol):
    # Set up the parameters for the API request
    parameters = {
        'function': 'TIME_SERIES_DAILY',
        'symbol': symbol,
        'apikey': API_KEY
    }
    
    # Make the HTTP GET request to the API
    response = requests.get(BASE_URL, params=parameters)
    data = response.json()

    # print("data=", data)
    
    # return data
    
    # # Extract the time series data
    if 'Time Series (Daily)' not in data:
        raise ValueError('Invalid API response. Check your symbol and API key.')
        
    ts_data = data['Time Series (Daily)']
    
    # Convert the JSON data into a pandas DataFrame for easier analysis
    df = pd.DataFrame.from_dict(ts_data, orient='index')
    df.columns = ['Open', 'High', 'Low', 'Close', 'Volume']
    df.index = pd.to_datetime(df.index)
    df = df.astype(float)
    
    return df

# Fetch and print the data
try:
    stock_df = fetch_stock_data(SYMBOL)
    print(stock_df.head())
except Exception as e:
    print(f"Error encountered: {e}")
```

```python
import requests
import pandas as pd
from datetime import datetime, timedelta

API_KEY = '5280A66MWDO73OD0'
BASE_URL = 'https://www.alphavantage.co/query'
SYMBOL = 'AAPL'

def fetch_one_year_data(symbol):
    parameters = {
        'function': 'TIME_SERIES_DAILY',
        'symbol': symbol,
        'outputsize': 'full',  # Request all data first
        'apikey': API_KEY
    }
    
    response = requests.get(BASE_URL, params=parameters)
    data = response.json()
    
    if 'Time Series (Daily)' not in data:
        raise ValueError('Invalid API response. Check your symbol and API key.')
        
    ts_data = data['Time Series (Daily)']
    
    # Convert to pandas DataFrame
    df = pd.DataFrame.from_dict(ts_data, orient='index')
    df.columns = ['Open', 'High', 'Low', 'Close', 'Volume']
    df.index = pd.to_datetime(df.index)
    df = df.astype(float)
    
    # Sort the index so the oldest dates are first
    df = df.sort_index()
    
    # Filter the DataFrame to only include data from the last 365 days
    one_year_ago = datetime.now() - timedelta(days=365)
    df_one_year = df[df.index >= one_year_ago]
    
    return df_one_year

# Fetch and print the filtered data
df_1y = fetch_one_year_data(SYMBOL)
print(df_1y)
```
